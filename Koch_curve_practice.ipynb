# Etapa A: Fundamento teórico

***
***
## En esta práctica necesitaremos las siguientes herramientas geométricas:
Fórmula del perímetro de un triángulo equilátero de lado $l$
$$P=3l$$
Fórmula del área de un triángulo equilátero de lado $l$
$$A=\frac{b\times h}{2}$$
donde $b$ es la base, en este caso $b=l$ y $h$ es la altura del triángulo, en este caso 
$$h=\frac{\sqrt{3}}{2}l$$
por lo que el área del triángulo equilátero de lado $l$ es:
$$A=\frac{l^{2}\sqrt{3}}{4}\tag{1}$$

## En esta práctica necesitaremos las siguientes herramientas de series:

Serie geométrica
Recuerda que una serie geométrica es aquella de la forma:
$$S=\sum_{n=0}^{\infty}r^n=1+r+r^2+r^3+\ldots+r^n+\ldots$$
Donde $r$ es la razón, recordemos que la serie geométrica se puede calcular por medio de:
$$S=\sum_{n=0}^{\infty}r^n=\lim_{n\to\infty}\frac{1-r^{n+1}}{1-r}$$
El límite de la derecha existe si $\vert r\vert<1$ y su valor es:
$$S=\sum_{n=0}^{\infty}r^n=\lim_{n\to\infty}\frac{1-r^{n+1}}{1-r}=\dfrac{1}{1-r}$$



```python
def koch_curve(order, size):
    if order == 0:
        t.forward(size)
    else:
        size /= 3.0
        koch_curve(order - 1, size)
        t.left(60)
        koch_curve(order - 1, size)
        t.right(120)
        koch_curve(order - 1, size)
        t.left(60)
        koch_curve(order - 1, size)
```

## Perímetro

Ahora calculemos el perímetro de la figura resultante, notemos que cada segmento es de longitud $1/3$ y hay 12 segmentos (o 12 lados), es decir, que por cada lado de la figura original ahora tenemos 4 segmentos nuevos ($N_1= 12 = 4*3 = 4*N_0$). Por lo tanto el perímetro total es:
$$P_1 = 12*\frac{1}{3} =  3*\left(\dfrac{4}{3}\right)= 4 = P_0 + 1 $$
## Área
Cómo segundo paso, calculemos el área de la nueva figura, observamos que el área de la nueva figura es el área de la figura anterior más el área de tres triángulos equiláteros de lado $1/3$.
El área de cada triángulo que hemos añadido la podemos calcular por medio de la ecuación $\eqref{eq_area}$ con lado $1/3$:
$$ A = \dfrac{\left(\dfrac{1}{3}\right)^2\sqrt{3}}{4} = \dfrac{\sqrt{3}}{4*3^2} = \dfrac{1}{9}*\dfrac{\sqrt{3}}{4} = \dfrac{1}{9} A_0$$
Como se han añadido tres triángulos, el área total es:
$$A_1 = A_0+3*\dfrac{A_0}{9} = A_0 \left(1+\dfrac{1}{3}\right) = A_0*\dfrac{3}{4}$$

```python
order = 2   # Nivel de detalle (puedes cambiarlo a 3, 4 o 5)
size = 100  # Longitud del lado
# 1. Crear el lienzo de la tortuga
myCanvas = turtle.Canvas(width=300, height=300)
display(myCanvas)
myTS=turtle.TurtleScreen(myCanvas)
myTS.clear()

t = turtle.Turtle(myTS)
#bob.shape("turtle")
salida_grafica = widgets.Output()
# 2. Crear el botón interactivo
boton_mover = widgets.Button(
    description="Avanzar y Girar",
    button_style="success",  # Color verde
    icon="play",
)
# Mover a un punto sin dejar rastro (opcional)
#t.penup()        # Levanta el lápiz para no dibujar la línea de transición
#t.goto(100, -50) # Mueve la tortuga exactamente a X=100, Y=-50
#t.pendown() 
#myTS.delay(200)
# 3. Definir la función que se ejecutará al hacer clic
def ejecutar_paso(b):
    # La tortuga avanza 50 píxeles y gira 45 grados en cada clic
    with salida_grafica:
        t.color("black", "lightblue") 
        t.begin_fill()
        for _ in range(3):
            koch_curve(order, size)
            t.right(120)
    t.end_fill()
#for _ in range(3):
#    koch_curve(order, size)
#    t.right(120)


# 4. Vincular el botón con la función
boton_mover.on_click(ejecutar_paso)

# 5. Organizar los elementos en un contenedor vertical (VBox)

# Esto es clave para que Voilà acomode la interfaz de forma ordenada
interfaz = widgets.VBox([salida_grafica, boton_mover])


# Mostrar la interfaz en el notebook / Voilà
interfaz

```

## Perímetro
Notemos lo siguiente: nuevamente por cada segmento de la iteración anterior se incorporan $4$ nuevos segmentos a la iteración actual, por lo que el total de segmentos en esta nueva iteración es 
$$N_2=4*12 = 4*3*4 = 3*4^2=N_1*4$$ 
Cada uno de estos segmentos tiene longitud $1/3$ del segmento anterior (que a su vez era $1/3$ de la longitud original) la nueva longitud de cada segmento es:
$$ l_2 = \dfrac{1}{9}=\dfrac{1}{3^2}$$ 
así, que el perímetro de la figura es:
$$P_2 = 3*4^2*\dfrac{1}{3^2} = 3*\left(\dfrac{4^2}{3^2}\right) = 3*\left(\dfrac{4}{3}\right)^2= \dfrac{16}{3} = 4+\dfrac{4}{3} = P_1+\dfrac{4}{3}= P_0+1+\dfrac{4}{3}$$
## Área
Calculemos el área de uno de los triángulos que hemos añadido. El área de cada triángulo que hemos añadido la podemos calcular por medio de la ecuación $\eqref{eq_area}$ con lado $1/9$:
$$A = \dfrac{\left(\dfrac{1}{9}\right)^2\sqrt{3}}{4} = \dfrac{\sqrt{3}}{4*9^2} = \dfrac{1}{9^2}*\dfrac{\sqrt{3}}{4} = \left(\dfrac{1}{9}\right)^2 A_0$$
Nuevamente tenemos al área de la figura anterior más el área de los nuevos triángulos que hemos añadido, en este caso se han añadido 12 triángulos nuevos, por lo que el área total es:
$$ A_2 = A_1 + 12*A = A_1 + 3*4*\left(\dfrac{1}{9}\right)^2 A_0$$
Recordemos que
$$A_1 = A_0\left(1+\dfrac{1}{3}\right)$$
podemos sustituir y obtener una expresión para $A_2$ solo en términos de $A_0$.

$$A_2 = A_0 \left( 1+\dfrac{1}{3}\right)+3*4*\left(\dfrac{1}{9}\right)^2 A_0$$

$$A_2 = A_0 \left( 1+\dfrac{3}{4}*\left(\dfrac{4}{9}\right)\right)+\dfrac{3}{4}*\left(\dfrac{4}{9}\right)^2 A_0$$

forma final:
$$A_2 = A_0 \left( 1+\dfrac{3}{4}*\left(\dfrac{4}{9}\right)+\dfrac{3}{4}*\left(\dfrac{4}{9}\right)^2 \right)$$

## Con los resultados de la sección anterior cálcula lo que se te pide

* si necesitas usar el infinito teclea: oo
* en caso de valores numéricos redondea a 3 decimales tu respuesta, por ejemplo: si tu resultado es 3.672890 este número se redondea a 3.363, 

```python
import sympy as sp
from sympy import symbols, limit, oo, sqrt
import ipywidgets as widgets
from IPython.display import display, clear_output, Markdown

# Inicializar impresión estética de SymPy en el cuaderno
sp.init_printing()
n = sp.symbols('n')

# 1. Cálculos matemáticos en el fondo
P = 3 * (4/3)**n
ecuacion_p = sp.sympify(P)
lim_perimetro = limit(ecuacion_p, n, oo) # Resultado: oo

A = ((sqrt(3)/4)/(5)) * (8 - 3*(4/9)**n)
ecuacion_a = sp.sympify(A)
lim_area = limit(ecuacion_a, n, oo)
lim_area_num = f"{lim_area.evalf():.3f}" # Resultado: 0.693

# 2. Creación de los componentes de la interfaz (Widgets)
lbl_instrucciones = widgets.HTML("<h2>Resultados finales de la práctica 4</h2><p>Calcula los límites cuando <b>n → ∞</b> e ingresa tus respuestas:</p>")

# Sección Perímetro
lbl_p = widgets.HTML("<h4><b>[ Caso del Perímetro ]</b></h4>")
txt_perimetro = widgets.Text(placeholder="Ej. 45.67 o infinito", description="Límite P:")

# Sección Área
lbl_a = widgets.HTML("<h4><b>[ Caso del Área ]</b></h4>")
txt_area = widgets.Text(placeholder="Ej. 12.256", description="Límite A:")

# Botón y Salidas
btn_verificar = widgets.Button(description="Verificar y Guardar", button_style="success", icon="check")
output_resultados = widgets.Output()

def procesar_practica(b):
    with output_resultados:
        clear_output()
        
        # Obtener y limpiar las respuestas del usuario
        resp_p = txt_perimetro.value.strip().lower()
        resp_a = txt_area.value.strip()
        
        if not resp_p or not resp_a:
            print("❌ Por favor, responde a ambos límites antes de verificar.")
            return
            
        # Validar Perímetro (Acepta 'oo' o 'infinito')
        p_correcto = resp_p in ['oo', 'infinito', '∞']
        
        # Validar Área
        a_correcto = resp_a == lim_area_num
        
        # Mostrar retroalimentación inmediata en pantalla usando Markdown
        display(Markdown("### 📊 Resultados de la Validación"))
        
        if p_correcto:
            display(Markdown("✅ **Perímetro:** ¡Felicidades, bien calculado el perímetro!"))
        else:
            display(Markdown(f"❌ **Perímetro:** Incorrecto. El valor correcto es `{lim_perimetro}` (infinito)."))
            
        if a_correcto:
            display(Markdown("✅ **Área:** ¡Felicidades, bien calculada el área!"))
        else:
            display(Markdown(f"❌ **Área:** Incorrecto. El valor correcto aproximado es `{lim_area_num}`."))
            
        # Escribir los resultados en el archivo de texto
        try:
            with open('Practica4.txt', 'a', encoding='utf-8') as f:
                f.write('\n' + 'Resultados finales de la práctica 4\n')
                f.write("************************************\n")
                f.write(f"En el caso del perímetro\n")
                f.write(f"El valor del límite ingresado fue: {resp_p}\n")
                if p_correcto:
                    f.write("Felicidades, bien calculado el perímetro\n")
                else:
                    f.write(f"El valor correcto es {lim_perimetro}\n")
                    
                f.write("************************************\n")
                f.write(f"En el caso del área\n")
                f.write(f"El valor del límite ingresado fue: {resp_a}\n")
                if a_correcto:
                    f.write("Felicidades, bien calculada el área\n")
                else:
                    f.write(f"El valor correcto es {lim_area_num}\n")
                f.write("************************************\n")
                
            print("\n💾 ¡Resultados respaldados exitosamente en 'Practica4.txt'!")
        except Exception as e:
            print(f"❌ Error al escribir en el archivo: {e}")

# Conectar el botón con la función reguladora
btn_verificar.on_click(procesar_practica)

# Diseñar la estructura visual en la pantalla
interfaz_mate = widgets.VBox([
    lbl_instrucciones,
    widgets.HTML("<hr>"),
    lbl_p, txt_perimetro,
    widgets.HTML("<br>"),
    lbl_a, txt_area,
    widgets.HTML("<br>"),
    btn_verificar,
    widgets.HTML("<br>"),
    output_resultados
])

display(interfaz_mate)
```
